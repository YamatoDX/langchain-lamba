```python
import os
api_key = os.getenv("OPENAI_API_KEY")
from dotenv import load_dotenv
load_dotenv()

from langchain_openai import ChatOpenAI

llm = ChatOpenAI(model="gpt-5-mini", temperature=0, api_key=api_key)
```

```python
from langchain_community.document_loaders import DirectoryLoader, PyPDFLoader

loader = DirectoryLoader(
    path="./pdf_documents",
    glob="*.pdf",
    loader_cls=PyPDFLoader
)

# 2. Load all documents into memory
docs = loader.load()
```

```python
for each in docs:
    print(each.metadata["source"])
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50
)
chunks = splitter.split_documents(docs)
```

```python
from langchain_openai import OpenAIEmbeddings
embedding_model = OpenAIEmbeddings()
```

```python
from langchain_community.vectorstores import Chroma 
vectorstore = Chroma.from_documents(
    documents=chunks,
    embedding=embedding_model,
    persist_directory="./Vector"
)
```

```python
from langchain_community.vectorstores import Chroma
vectorstore = Chroma(
    persist_directory="./Vector",
    embedding_function=embedding_model
)
```

```python
context = vectorstore.similarity_search("All the information about md shadman matin")
for index, each in enumerate(context):
    print(f'{index+1}. >> {each.metadata["source"]}')
    print(each.page_content)
    print("===============================================")
```

```python
response = llm.invoke(f'what is md shadman matin BMET information within this context: {context}')
print(response.content)
```
