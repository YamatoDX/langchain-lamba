# RAG APPLICATION with PDF

```python
import os
api_key = os.getenv("OPENAI_API_KEY")
from dotenv import load_dotenv
load_dotenv()

from langchain_openai import ChatOpenAI
```

```python
llm = ChatOpenAI(model="gpt-5-mini", temperature=0, api_key=api_key)
```

```python
from langchain_community.document_loaders import PyPDFLoader

loader = PyPDFLoader("./TIN.pdf")
docs = loader.load()
```

```python
docs[0].page_content
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50
)
chunks = splitter.split_documents(docs)
```

```python
from langchain_openai import OpenAIEmbeddings
embedding_model = OpenAIEmbeddings()
```

```python
from langchain_community.vectorstores import Chroma 
vectorstore = Chroma.from_documents(
    documents=chunks,
    embedding=embedding_model
)
```

```python
print(vectorstore)
```

```python
context = vectorstore.similarity_search("INFORMATION ABOUT MAHIN UL HUQUE MRIDHA")
response = llm.invoke(f'What will be the penalty of not filing tax within this context:{context}')
print(response.content)
```
